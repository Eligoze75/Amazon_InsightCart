## Conclusions

We decided to drop the following columns:
- **reviews**
    - images: We won't be retrieving or analyzing images in our project. That would be out of the scope, so we can drop it.
    - asin: This ID is not useful to do any joins (we use `parent_asin` to do the join with metadata). Since it has no use, we can drop it. The tradeoff could be that the agent won't be able to give answers at asin level.
    - user_id: We have no use case for this column. Queries on users will be not possible, but the loss of this case doesn't hurt.
    - helpful_vote: This column is unary, so it doesn't really bring a lot of value.
    - verified_purchase: This column is also unary, therefore it doesn't really bring a lot of value.
    
- **metadata**
    - main_category: This column is redundant. We already know that the category is Video Games.
    - bought_together: This column is completely or almost completely null, so we decided to drop it.
    - subtitle: This column is completely or almost completely null, so we decided to drop it.
    - author: This column is completely or almost completely null, so we decided to drop it.
    - images: We won't be retrieving or analyzing images in our project, nor their urls. That would be out of the scope, so we can drop it.
    - videos: We won't be retrieving or analyzing videos in our project, nor their urls. That would be out of the scope, so we can drop it.

Then, we also noticed that the `main_category` in the metadata dataset is not always video games and that many cases are not very video games related, like the examples we found of _"Buy a Kindle"_.
